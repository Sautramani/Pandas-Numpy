```python
# CONVERTERS (converters ka use CSV se data read karte time kisi particular column ke har value par apna function apply karne ke liye hota hai.)
def rename(name):
    if name == "Delhi Daredevils":
      return "DC"
    if name == "Royal Challengers Bangalore":
      return "RCB"
    else:
        return name
```

```python
rename("Royal Challengers Bangalore")
```

```python
pd.read_csv('ipl-matches.csv',converters={'Team1':rename})
```

```python
#LOADING A HUGE DATASET IN CHUNKS
dfs=pd.read_csv('aug_train.csv',chunksize=5000)
for chunks in dfs:
  print(chunks)
```

```python
# Using a for loop and enumerate to print only the 3rd chunk (index 2)
dfs = pd.read_csv('aug_train.csv', chunksize=5000)

for index, chunk in enumerate(dfs):
    if index == 2:
        display(chunk)
        break
```

# EXPORT DATASET AS CSV FILE

```python
df=pd.read_csv('ipl-matches.csv')
df=df.groupby('Team1')['Margin'].sum().reset_index()
df.to_csv('neww.csv',index=False)#AB NEW CSV FILE BAN GAYI PURANI ipl-matches WALI CSV FILE KI HELP SE
pd.read_csv('neww.csv')
```

# EXPORT DATASET AS EXCEL **FILE**

```python
df=pd.read_csv('ipl-matches.csv')
df=df.groupby('Team1')['Margin'].sum().reset_index()
df.to_excel('new2.xlsx')
df
```
